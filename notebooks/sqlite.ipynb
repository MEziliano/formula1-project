```python
# Scraping + Database

!pip install mechanicalsoup

import mechanicalsoup
import pandas as pd
import sqlite3

browser = mechanicalsoup.StatefulBrowser()
browser.open("https://en.wikipedia.org/wiki/Comparison_of_Linux_distributions#")
# extract table headers
th = browser.page.find_all("th", attrs={"class": "table-rh"})
distribution = [value.text for value in th]
print(distribution)

distribution = [value.text.replace("\n", "") for value in th]
print(distribution)

print(distribution.index("Zorin OS"))

distribution = distribution[:98]
print(distribution)

#extract table data
td = browser.page.find_all("td")
columns = [value.text.replace("\n", "") for value in td]
# print(columns.index("AlmaLinux Foundation"))
# print(columns.index("Binary blobs"))

columns = columns[7:1085]
print(columns)

# columns[::11]
# columns[1:][:11]
# conlumns[2::][:11]

colum_names = ["Founder",
               "Maintainer",
               "Initial_Release_year",
               "Current_Stable_Version",
               "Security_Updates", "Release_Date",
               "System_Distribution_Commitment",
               "Forked_from", "Target_Audience",
               "Cost", "Status"]

dictionary = {"Distribution": distribution}

for idx, key in enumerate(colum_names):
  dictionary[key] = columns[idx:][::11]
df = pd.DataFrame(data = dictionary)
print(df.head())

df.head()

connection = sqlite3.connect("linux_distro.db")
cursor = connection.cursor()
cursor.execute("create table linux (Distribution," + ",".join(colum_names) + ")")
for i in range(len(df)):
  cursor.execute("insert into linux values(?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?, ?)", df.iloc[i])
connection.commit()
connection.close()

import requests
from bs4 import BeautifulSoup
import pandas as pd

# URL da página da Wikipedia
url = "https://en.wikipedia.org/wiki/Comparison_of_Linux_distributions#"

# Realiza a requisição HTTP
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

# Encontre a tabela com os dados desejados
table = soup.find("table", {"class": "wikitable"})

# Inicializa as listas para armazenar os dados
distributions = []
release_models = []
initial_release_dates = []
latest_stable_release_dates = []

# Extrai os dados da tabela
for row in table.find_all("tr")[1:]:
    columns = row.find_all(["th", "td"])

    if len(columns) >= 4:
        distribution = columns[0].text.strip()
        release_model = columns[1].text.strip()
        initial_release_date = columns[2].text.strip()
        latest_stable_release_date = columns[3].text.strip()

        distributions.append(distribution)
        release_models.append(release_model)
        initial_release_dates.append(initial_release_date)
        latest_stable_release_dates.append(latest_stable_release_date)

# Cria um DataFrame usando pandas
data = {
    "Distribution": distributions,
    "Release Model": release_models,
    "Initial Release Date": initial_release_dates,
    "Latest Stable Release Date": latest_stable_release_dates
}
```

```python
df = pd.DataFrame(data)

# Exibe o DataFrame
print(df)
```
